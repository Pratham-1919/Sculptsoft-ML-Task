```python
import pandas as pd
```

```python
df = pd.read_csv(r"E:\sculptsoft\28 July\Twitter Sentiments.csv")
df.head()
```

### Performing text preprocessing

```python
df = df.drop(columns=["id","label"])
df.head()
```

```python
df["clean_text"] = df["tweet"].str.lower()
df.head()
```

```python
import string

string.punctuation
```

```python
def remove_punctuations(text):
    punctuations = string.punctuation
    return text.translate(str.maketrans("","",punctuations))
```

```python
df["clean_text"] = df["clean_text"].apply(lambda x : remove_punctuations(x))
df.head()
```

```python
import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords
stop_words = stopwords.words("english")
print(stop_words)
```

```python
stop_words = stopwords.words("english")
def remove_stop_words(text):
    return " ".join([word for word in text.split() if word not in stop_words])
```

```python
df["clean_text"] = df["clean_text"].apply(lambda x : remove_stop_words(x))
df.head(10)
```

### Count the most frequent word and remove them

```python
from collections import Counter

word_count = Counter()

for text in df["clean_text"]:
    for word in text.split():
        word_count[word] +=1

word_count.most_common(10)
```

```python
frequent_word = set(word for (word,wc) in word_count.most_common(10))

def remove_frequent_word(text):
    return " ".join([word for word in text.split() if word not in frequent_word])
```

```python
df["clean_text"] = df["clean_text"].apply(lambda x : remove_frequent_word(x))
df.head(10)
```

### Removal of special character

```python
import re

def remove_character(text):
    text = re.sub("[^a-zA-Z0-9]", " ", text)
    return text
```

```python
df["clean_text"] = df["clean_text"].apply(lambda x : remove_character(x))
df.head(10)
```

### performing stemming

```python
from nltk.stem.porter import PorterStemmer

ps = PorterStemmer()
def stem_words(text):
    return " ".join([ps.stem(word) for word in text.split()])
```

```python
df["stemming"] = df["clean_text"].apply(lambda x : stem_words(x))
df.head()
```

### Removal of url in a text

```python
text = "https://www.pratham_langaliya.com this is the url of my website. Please go through it and subscribe my channel"

```

```python
def removal_url(text):
    return re.sub(r'https://\S+|www\.\S+',' ',text)
```

```python
print(removal_url(text))
```
